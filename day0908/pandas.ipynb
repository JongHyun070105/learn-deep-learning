## 2. indexing과 slicing

- DataFrame.loc: 행과 열의 이름을 사용
- DataFrame.iloc: 행과 열의 정수형 index를 사용
- DataFrame[행 index, 열 index]의 형태로 사용할 수 없으며, 반드시 loc 또는 iloc를 사용해야함
- 단, DataFrame[열이름]으로 특정 열을 추출할 수 있음


```python
# Ben의 연봉(Salary)

# df['Ben', 'Salary'] 안됨

df.loc['Ben', 'Salary'] # 대부분 loc 씀
df.iloc[8,3] # 인덱싱할 때 이거 잘 안씀
```

```python
# Tim부터 Linda까지, Age부터 Salary까지
# loc를 사용할 경우, a:b는 a부터 b까지를 의미하며, b가 포함됨에 주의할 것
# iloc를 통해 정수형 index a:b 범위를 지정할 때, index b는 제외됨에 주의할 것
# 범위가 아닌 복수의 열 또는 행의 이름으로도 지정 가능

# df.loc[['Tim', 'Linda', 'Emma'], ['Age','Salary']]
# df.iloc[2:7,1:3]
df.loc['Tim':'Linda', 'Age':'Salary']
```

- DataFrame.drop([col1, col2, ...], axis = 1): 지정된 열을 삭제
- DataFrame.drop([col1, col2, ...], axis = 0): 지정된 행을 삭제
- col1, col2를 index로 axis = 0으로 대체할 경우, 행을 삭제
